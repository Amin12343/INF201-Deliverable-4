## Task 1
In this task we create a Temperature class in temperature.py that supports Celsius, Fahrenheit and Kelvin. The value and unit are read-only, invalid units and temperatures below absolute zero raise a ValueError, and all six comparisons work across units with a tolerance of 1e-9.

```python
from temperature import Temperature 

room = Temperature(21.5)  # no unit given, so Celsius is used
body = Temperature(98.6, "f")  # lower-case unit is accepted
space = Temperature(2.7, "K")  # temperature of outer space

print(repr(room), repr(body), repr(space))  # unambiguous representation
print(room, body, space)  # human-readble with two decimals
print(f"Value: {room.value}, unit: {room.unit}")  # read-only properties
```

```python
# convert between units 
print(f"{room} = {room.to_fahrenheit():.2f} °F = {room.to_kelvin():.2f} K") #from celsius 
print(f"{body} = {body.to_celsius():.2f} °C = {body.to_kelvin():.2f} K")  # from Fahrenheit
body_in_celsius = body.convert_to("C")  # new Temperature object iin Celsius
print(repr(body_in_celsius), body_in_celsius)
print(repr(body))  # teh original object is not changed
```

```python
#compare temperatures in different units 
freezing_c = Temperature(0, "C") #freezing point in Celsius 
freezing_f = Temperature(32, "F")  # freezing point in Fahrenheit
print(f"{freezing_c} == {freezing_f}: {freezing_c == freezing_f}")
print(f"{room} < {body}: {room < body}")
print(f"{space} >= {freezing_c}: {space >= freezing_c}")
print(f"{freezing_c} == 0: {freezing_c == 0}")  # not a Temperature, so False
```

```python
# Show the error messages for invalid input
for value, unit in [(25, "X"), (-300, "C"), (-1, "K")]:  # invalid unit and below absolute zero
    try:
        Temperature(value, unit)  # this raises a ValueError
    except ValueError as err:
        print(f"ValueError: {err}")

try:
    room.value = 30  # value is read-only
except AttributeError as err:
    print(f"AttributeError: {err}")
```
